## Data Cleaning 

### Standardising the Item Fat Content Column by cleaning the category value names

```python
df['Item Fat Content'] = df['Item Fat Content'].replace({"LF" : "Low Fat", "low fat" : "Low Fat" , "reg" : "Regular"})
```

```python
df['Item Fat Content'].unique()
```

## Business Requirements

```python
# Total Sales

Total_Sales = df['Sales'].sum()

# Average Sales

Average_Sales = df['Sales'].mean()

#Number of Items Sold

Items_Sold = df['Sales'].count()

#Average Ratings 

Average_Rating = df['Rating'].mean()

print(f"Total Sales : ${Total_Sales:,.1f} \nAverage Sales : ${Average_Sales:,.1f} \nTotal Number of Items Sold : {Items_Sold} \nAverage Rating : {Average_Rating:,.1f}")
```

### Total Sales by FAT content

```python
SalesByFatContent = df.groupby("Item Fat Content")["Sales"].sum()
plt.pie(SalesByFatContent , autopct = "%1.1f%%" , labels=SalesByFatContent.index)
plt.title("Total Sales by Fat Content Distribution")
plt.legend()
plt.tight_layout()
plt.show()
```

### 📊 Observations: Total Sales by Fat Content
* **Health-Conscious Preferences:** The clear majority of sales are driven by **Low Fat** items, which generated approximately **64.6%** compared to **35.4%** for Regular items. 
* **Actionable Insight:** Consumers are showing a strong preference for healthier options. Inventory planning and marketing campaigns should prioritize expanding and promoting the Low Fat product catalog to maximize revenue.

### Total Sales By Item Type

```python
ItemTypeBySales = df.groupby("Item Type")["Sales"].sum().sort_values(ascending=False)
plt.title("Total Sales By Item Type")
ax = ItemTypeBySales.plot(kind="barh")

for container in ax.containers:
    custom_labels = [f"${value:,.1f}" for value in container.datavalues]
    ax.bar_label(container,padding=3 , labels=custom_labels)

plt.ylabel("Types of Items")
plt.xlabel("Total Sales")
plt.xticks(rotation=45)

left,right = ax.get_xlim()
ax.set_xlim(left,right*1.2)

plt.tight_layout()
plt.show()
```

### 📊 Observations: Total Sales by Item Type
* **Top Performers:** Revenue is heavily concentrated at the top, with **Fruits and Vegetables** ($178,124) and **Snack Foods** ($175,434) being the highest-grossing categories. Together, fresh produce and quick snacks drive the core of the business.
* **Underperformers:** Categories like **Seafood** ($9,078) and **Breakfast** ($15,597) yield the lowest sales. 
* **Conclusion:** The store operates primarily as a destination for daily essentials and snacks. Lower-performing categories might represent niche markets; we should evaluate whether they suffer from low stock availability or simply lack consumer demand.

### FAT content for outlet by Total Sales

```python
FatContentForOutlet = df.groupby(["Outlet Location Type","Item Fat Content"])["Sales"].sum().unstack()
ax = FatContentForOutlet.plot(kind="bar" , title="Fat Content for outlet by Total Sales Distribution")

for container in ax.containers:
    custom_labels = [f"${value:,.1f}" for value in container.datavalues]
    ax.bar_label(container,padding=3,labels=custom_labels)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 📊 Observations: Fat Content Sales Across Outlet Tiers
* **Consistent Fat Content Trends:** Regardless of the outlet location type (Tier 1, Tier 2, or Tier 3), **Low Fat** products consistently outsell Regular products by a significant margin. The health-conscious trend is universal across all store demographics.
* **Tier 3 Dominance:** **Tier 3** locations generate the highest absolute sales for both Low Fat ($306,807) and Regular ($165,326) items, outperforming Tiers 1 and 2.
* **Conclusion:** Tier 3 locations are the most lucrative markets. Supply chain distribution should ensure these high-volume stores are consistently well-stocked, particularly with Low Fat inventory.

### Total Sales by Outlet Establishment

```python
EstablishmentSales = df.groupby("Outlet Establishment Year")["Sales"].sum()
ax = EstablishmentSales.plot(kind="line" , title ="Total Sales by Outlet Establishment Year" ,figsize=(12,6) , legend=True , marker='o')

for x,y in EstablishmentSales.items():
    label = f"${y:,.0f}"
    ax.annotate(label,(x,y),textcoords="offset points",xytext=(0,10),ha="center",fontsize=9)

plt.ylabel("Sum of Sales")
plt.tight_layout()
plt.show()
```

### 📊 Observations: Total Sales by Outlet Establishment Year
* **Legacy Outlets Lead:** Outlets established in **1998** generated the highest total sales peak at approximately **$204,522**. This indicates that older, more established stores benefit from high customer loyalty and foot traffic.
* **Stable Recent Growth:** Stores established from 2012 to 2022 show highly consistent sales figures, all hovering around the **$130,000** mark.
* **Notable Dip:** Outlets established in **2011** are a major outlier, generating significantly lower sales ($78,131). This specific cohort of stores warrants further investigation to understand why they are underperforming compared to both older and newer stores.

### Total Sales by Outlet Size

```python
SalesbyOutlet = df.groupby("Outlet Size")["Sales"].sum()
plt.pie(SalesbyOutlet , autopct="%1.1f%%" , labels=SalesbyOutlet.index , startangle=90)
plt.title("Total Sales by Outlet Size Distribution")
plt.legend(title="Outlet Size")
plt.tight_layout()
plt.show()
```

### 📊 Observations: Total Sales by Outlet Size Distribution
* **Medium Outlets Lead:** Medium-sized outlets are the primary revenue drivers, accounting for **42.3%** of total sales. 
* **Small vs. High:** Small outlets also perform strongly at **37.0%**, while 'High' (large) outlets contribute the smallest share at **20.7%**.
* **Actionable Insight:** Expansion or inventory optimization efforts should focus heavily on Medium and Small format stores, as they represent nearly 80% of the total sales volume. Larger format stores may need a strategic review to improve their sales density.

### Total Sales by Outlet Location Type

```python
SalesbyOutletLocation = df.groupby("Outlet Location Type")["Sales"].sum().reset_index()
ax = sns.barplot(x="Sales",y="Outlet Location Type" , data=SalesbyOutletLocation)
ax.bar_label(ax.containers[0], padding=3 , labels=[f"${value:,.1f}" for value in ax.containers[0].datavalues])
left,right = ax.get_xlim()
ax.set_xlim(left,right*1.2)
plt.title("Total Sales by Outlet Location Type")
plt.xlabel("Total Sales")
plt.ylabel("Outlet Location Type")
plt.tight_layout()
plt.show()
```

### 📊 Observations: Total Sales by Outlet Location Type
* **Tier 3 Dominance:** **Tier 3** locations generate the highest total sales by a wide margin at **$472,133.0**, making them the most lucrative market segment.
* **Tier 2 & 1 Performance:** **Tier 2** locations follow with **$393,150.6**, while **Tier 1** locations generate the lowest sales among the groups at **$336,397.8**.
* **Actionable Insight:** Tier 3 and Tier 2 areas are significantly outperforming Tier 1 locations. Business strategy should prioritize supply chain efficiency, inventory expansion, and marketing in these highly profitable, lower-tier locations.
